```julia
# Create color gradients with your colors in the middle
function create_gradient_colors(base_color::RGBA, n_colors::Int)
    hsv_color = convert(HSV, base_color)
    colors = RGBA[]
    
    middle = (n_colors + 1) / 2
    
    for i in 1:n_colors
        if i < middle
            # Lighter colors: go most of the way to white
            t = i / middle  # 0 to 1
            new_v = hsv_color.v + (1.0 - hsv_color.v) * 0.75 * (1 - t)  # 75% toward white
            new_s = hsv_color.s * (0.25 + 0.75 * t)  # Keep minimum 25% saturation
        else
            # Darker colors: darken substantially
            t = (i - middle) / (n_colors - middle)  # 0 to 1
            new_v = hsv_color.v * (1 - 0.5 * t)  # 50% darker
            new_s = hsv_color.s
        end
        
        new_hsv = HSV(hsv_color.h, clamp(new_s, 0.0, 1.0), clamp(new_v, 0.0, 1.0))
        push!(colors, convert(RGBA, new_hsv))
    end
    
    return colors
end
```

# DC sweep simu

```julia
# Generate gradient colors
blue_colors = create_gradient_colors(myBlue, 100)
purple_colors = create_gradient_colors(myPurple, 100)

iDC_vals_normal, outputs_DC_normal = get_parameter_sweep_data("./Cadence_output/normal.vcsv")
Igain = get_all_xy_pairs("./Cadence_output/Igain.vcsv")
Iwidth = get_all_xy_pairs("./Cadence_output/Iwidth.vcsv")

# Now plot with the gradients
panelB = plot(legend=false, margins=20Plots.px, size=(1200, 800))

for i = 1:100
    x, y = Igain[i] 
    plot!(x .* 1e12, y * 1e12, linewidth=4., color=blue_colors[i], alpha=0.5, label=false)
end

for i = 1:100
    x, y = Iwidth[i] 
    plot!(x .* 1e12, y * 1e12, linewidth=4., color=purple_colors[i], alpha=0.5, label=false)
end

plot!(iDC_vals_normal .* 1e12, outputs_DC_normal * 1e12, linewidth=6., color=:black, legend=:topleft,
    margins=20Plots.px, size=(1200, 800), label=false)
xlabel!(L"I_\mathrm{in}\, (\mathrm{pA})")
ylabel!(L"I_\mathrm{out}\, (\mathrm{pA})")
plot!(yticks=[0, 100, 200, 300, 400, 500, 600], xticks=[0, 100, 200, 300, 400, 500])
xlims!((0, 500))
ylims!((0, 600))

display(panelB)
savefig(panelB, "./Cadence_plots/panelB.pdf")
```
