# Parameter sweeps simus

```julia
# Get transient data
igain_vals, outputs_gain = get_parameter_sweep_data("./Cadence_output/bistable_cell_Igain_sweep.vcsv")

sweep_gain = plot(igain_vals .* 1e12, outputs_gain * 1e12, linewidth=4., color=1, legend=:topleft,
    margins=20Plots.px, size=(1200, 800), label=L"\mathrm{Simulated}")

sweep_th(i) = i
plot!(igain_vals .* 1e12, sweep_th.(igain_vals) * 1e12, linewidth=4., color=myGray, legend=:topleft,
    margins=20Plots.px, size=(1200, 800), label=L"\mathrm{Theoretical}")

xlabel!(L"I_\mathrm{gain}\, (\mathrm{pA})")
ylabel!(L"\mathrm{High}\,\, I_\mathrm{out}\,(\mathrm{pA})")
plot!(yticks=[0, 100, 200, 300, 400, 500], xticks=[0, 100, 200, 300, 400, 500])

display(sweep_gain)
# savefig(sweep_gain, "./figures_pdf/sweep_gain_grid.pdf")
```

```julia
# Get parameter sweep data 
ithresh_vals, outputs_thresh = get_parameter_sweep_data("./Cadence_output/bistable_cell_Ithresh_sweep.vcsv")

sweep_thresh = plot(ithresh_vals .* 1e12, outputs_thresh * 1e12, linewidth=4., color=1, 
    legend=:topleft, margins=20Plots.px, size=(1200, 800), label=L"\mathrm{Simulated}")

sweep_th(i) = i
plot!(ithresh_vals .* 1e12, sweep_th.(ithresh_vals) * 1e12, linewidth=4., color=myGray, legend=:topleft,
    margins=20Plots.px, size=(1200, 800), label=L"\mathrm{Theoretical}")

xlabel!(L"I_\mathrm{thresh}\, (\mathrm{pA})")
ylabel!(L"\mathrm{High}\,\, \mathrm{threshold}\,(\mathrm{pA})")
plot!(yticks=[100, 200, 300, 400], xticks=[100, 200, 300, 400])

display(sweep_thresh)
# savefig(sweep_thresh, "./figures_pdf/sweep_thresh_grid.pdf")
```

```julia
# Get parameter sweep data  
iwidth_vals, outputs_width = get_parameter_sweep_data("./Cadence_output/bistable_cell_Iwidth_sweep.vcsv")

sweep_width = plot(iwidth_vals .* 1e12, outputs_width * 1e12, linewidth=4., color=1, 
    legend=:topleft, margins=20Plots.px, size=(1200, 800), label=L"\mathrm{Simulated}")
sweep_th(i) = i
plot!(iwidth_vals .* 1e12, sweep_th.(iwidth_vals) * 1e12, linewidth=4., color=myGray, legend=:topleft,
    margins=20Plots.px, size=(1200, 800), label=L"\mathrm{Theoretical}")

xlabel!(L"I_\mathrm{width}\, (\mathrm{pA})")
ylabel!(L"\mathrm{Hysteresis}\,\, \mathrm{width}\,(\mathrm{pA})")
plot!(yticks=[0, 100, 200, 300], xticks=[0, 100, 200, 300])
xlims!((-7, 307))

display(sweep_width)
# savefig(sweep_width, "./figures_pdf/sweep_width_grid.pdf")
```

```julia
# Create an interpolation from y to x (for inverse prediction)
# First, make sure y is sorted (required for interpolation)
sorted_indices1 = sortperm(outputs_gain)
y_sorted1 = outputs_gain[sorted_indices1]
x_sorted1 = igain_vals[sorted_indices1]

itp1 = linear_interpolation(y_sorted1, x_sorted1, extrapolation_bc=Line())
alpha_value = 0.5e-9
igain = itp1(alpha_value)

true_alpha_value = alpha_value*1e9
true_igain = igain*1e9
println("For alpha = $true_alpha_value, Igain = $true_igain nA")
```

```julia
# Create an interpolation from y to x (for inverse prediction)
# First, make sure y is sorted (required for interpolation)
sorted_indices2 = sortperm(outputs_thresh)
y_sorted2 = outputs_thresh[sorted_indices2]
x_sorted2 = ithresh_vals[sorted_indices2]

itp2 = linear_interpolation(y_sorted2, x_sorted2, extrapolation_bc=Line())
beta2_value = 0.4e-9
ithresh = itp2(beta2_value)

true_beta2_value = beta2_value*1e9
true_ithresh = ithresh*1e9
println("For beta2 = $true_beta2_value, Ithresh = $true_ithresh nA")
```

```julia
# Create an interpolation from y to x (for inverse prediction)
# First, make sure y is sorted (required for interpolation)
sorted_indices3 = sortperm(outputs_width)
y_sorted3 = outputs_width[sorted_indices3]
x_sorted3 = iwidth_vals[sorted_indices3]

itp3 = linear_interpolation(y_sorted3, x_sorted3, extrapolation_bc=Line())
delta_value = 0.2e-9
iwidth = itp3(delta_value)

true_delta_value = delta_value*1e9
true_iwidth = iwidth*1e9
println("For delta = $true_delta_value, Iwidth = $true_iwidth nA")
```
